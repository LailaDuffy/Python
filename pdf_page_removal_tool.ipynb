```python
from google.colab import files

uploaded = files.upload()  # Upload the PDF
pdf_filename = list(uploaded.keys())[0]  # Get the filename

```

```python
from pypdf import PdfReader, PdfWriter

# Read the uploaded PDF
reader = PdfReader(pdf_filename)
writer = PdfWriter()

# Add all pages except the last one
for i in range(len(reader.pages) - 1):
    writer.add_page(reader.pages[i])

# Save the new PDF
output_pdf = "modified_" + pdf_filename
with open(output_pdf, "wb") as output_file:
    writer.write(output_file)

# Download the modified file
files.download(output_pdf)

print(f"Last page removed. Download your file: {output_pdf}")
```

```python
from pypdf import PdfReader, PdfWriter

# Read the uploaded PDF
reader = PdfReader(pdf_filename)
writer = PdfWriter()

# Get the last page only
last_page = reader.pages[-1]
writer.add_page(last_page)

# Save the new PDF
output_pdf = "last_page_" + pdf_filename
with open(output_pdf, "wb") as output_file:
    writer.write(output_file)

# Download the modified file
files.download(output_pdf)

print(f"Only the last page is kept. Download your file: {output_pdf}")
```
