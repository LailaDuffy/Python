```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import datetime

df = pd.read_csv('/content/5a1ade24-42da-43cc-8d03-d5b27aa80b23.csv')

df
```

```python
# Convert lastPolicySync to datetime format
df['lastPolicySync'] = pd.to_datetime(df['lastPolicySync'], errors='coerce')

# Define the cutoff date
cutoff_date = datetime.datetime(2025, 1, 1)

# Determine if each device has been switched on since 01/01/2025
df['Status'] = df['lastPolicySync'].apply(lambda x: 'Lietoti kopš 2025. gada 1. janvāra' if x >= cutoff_date else 'Nav lietoti kopš 2025. gada 1. janvāra')

# Ensure annotatedLocation column has no extra spaces and remove NaN values
df['annotatedLocation'] = df['annotatedLocation'].astype(str).str.strip()

# Remove specific locations from the dataset completely
df = df[~df['annotatedLocation'].str.contains("Gulbenes novada Izglītības pārvalde|ITKPC", na=False)]

# Group by annotatedLocation and count statuses
school_status_counts = df.groupby(['annotatedLocation', 'Status']).size().unstack(fill_value=0)

# Plot stacked bar chart for each school
fig, ax = plt.subplots(figsize=(12, 6))
school_status_counts.plot(kind='bar', stacked=True, ax=ax, color=['tab:blue', 'tab:red'])

# Labels and title
ax.set_ylabel('Chromebook skaits')
ax.set_xlabel('Skola')
ax.set_title('Chromebook lietošana Gulbenes novada skolās')
ax.legend(title='Statuss')

# Rotate x-axis labels for better readability
plt.xticks(rotation=45, ha='right')

# Show the plot
plt.show()
```
